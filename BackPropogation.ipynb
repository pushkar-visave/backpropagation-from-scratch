# Backpropagation from Scratch vs Keras

A neural network (2 → 2 → 1) built manually using NumPy, then
replicated in Keras to compare predictions and loss.

Dataset : 4 students — CGPA, Profile Score → LPA (salary)

Network : Input(2) → Hidden(2) → Output(1)

Loss    : Mean Squared Error

Optimiser: SGD, learning rate = 0.001

Epochs  : 5

Sections:
1. Manual implementation (initialize → forward → backprop → train)
2. Keras equivalent
3. Comparison table

#Section 1- Manual Implementation

```python
import numpy as np
import pandas as pd
```

```python
# This is a regression problem: predict LPA from CGPA + profile score
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
df
```

```python
#In layer_dims = (# of input features, # of hidden layers, # of output layer)
#Initializing of weights for the first time
def initialize_parameters(layer_dims):

  np.random.seed(3)
  parameters = {}
  L = len(layer_dims)

  for l in range(1, L):

    parameters['W' + str(l)] = np.ones((layer_dims[l-1], layer_dims[l]))*0.1
    parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))


  return parameters

# It will select the values for weights and biases
```

```python
initialize_parameters([2,2,1])
```

```python
def linear_forward(A_prev, W, b):

  Z = np.dot(W.T, A_prev) + b

  return Z
```

```python
# Forward Propogation for first epoch
def L_layer_forward(X, parameters):

  A = X
  L = len(parameters) // 2                  # number of layers in the neural network

  for l in range(1, L+1):
    A_prev = A
    Wl = parameters['W' + str(l)]
    bl = parameters['b' + str(l)]

    A = linear_forward(A_prev, Wl, bl)
    #print("A"+str(l)+": ", A)
    #print("**"*20)

  return A,A_prev
```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1)
y = df[['lpa']].values[0][0]

# Parameter initialization
parameters = initialize_parameters([2,2,1])

y_hat,A1 = L_layer_forward(X, parameters)
```

```python
#To update the parameter after first epoch
def update_parameters(parameters,y,y_hat,A1,X):
  parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * (y - y_hat)*A1[0][0])
  parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat)*A1[1][0])
  parameters['b2'][0][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat))

  parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0])

  parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0])
```

```python
# Output after first epoch
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1)
y = df[['lpa']].values[0][0]

parameters = initialize_parameters([2,2,1])

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# Output after second epoch
X = df[['cgpa', 'profile_score']].values[1].reshape(2,1)
y = df[['lpa']].values[1][0]

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# Output after third epoch
X = df[['cgpa', 'profile_score']].values[2].reshape(2,1) # Shape(no of features, no. of training exaplme)
y = df[['lpa']].values[2][0]

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# Output after fourth epoch
X = df[['cgpa', 'profile_score']].values[3].reshape(2,1) # Shape(no of features, no. of training exaplme)
y = df[['lpa']].values[3][0]

y_hat,A1 = L_layer_forward(X,parameters)
y_hat = y_hat[0][0]

update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# epochs altogether

parameters = initialize_parameters([2,2,1])
epochs = 5

for i in range(epochs):

  Loss = []

  for j in range(df.shape[0]):

    X = df[['cgpa', 'profile_score']].values[j].reshape(2,1) # Shape(no of features, no. of training example)
    y = df[['lpa']].values[j][0]

    # Parameter initialization


    y_hat,A1 = L_layer_forward(X,parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters,y,y_hat,A1,X)

    Loss.append((y-y_hat)**2)

  print('Epoch -',i+1,'Loss -',np.array(Loss).mean())

parameters
```

#Section 2- Keras implementation

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers
```

```python
X = df[['cgpa','profile_score']].values   # inputs
y = df['lpa'].values
```

```python
X
```

```python
y
```

```python
model = keras.Sequential([keras.Input(shape=(2,)),
    layers.Dense(2, use_bias=True),
    layers.Dense(1, use_bias=True),
])
```

```python
n0_weight = np.ones((2,2))*0.1, np.zeros(2)
n0_weight
```

```python
n1_weight = np.ones((2,1))*0.1, np.zeros(1)
n1_weight
```

```python
model.layers[0].set_weights(n0_weight)
model.layers[1].set_weights(n1_weight)
```

```python
model.compile(optimizer=keras.optimizers.SGD(learning_rate=0.001),loss='mean_squared_error')
```

```python
history = model.fit(X, y, epochs=5, batch_size=1, verbose=1)
```

```python
print(model.predict(X))
```

# Section 3- Comparison table

```python
#Comparison Tables

parameters = initialize_parameters([2,2,1])
manual_loss_per_epoch = []

for i in range(5):
    Loss = []
    for j in range(df.shape[0]):
        Xj = df[['cgpa', 'profile_score']].values[j].reshape(2,1)
        yj = df[['lpa']].values[j][0]
        y_hat, A1 = L_layer_forward(Xj, parameters)
        y_hat = y_hat[0][0]
        update_parameters(parameters, yj, y_hat, A1, Xj)
        Loss.append((yj - y_hat)**2)
    manual_loss_per_epoch.append(np.array(Loss).mean())

# Final predictions from my model
manual_preds = []
for j in range(df.shape[0]):
    Xj = df[['cgpa', 'profile_score']].values[j].reshape(2,1)
    y_hat, _ = L_layer_forward(Xj, parameters)
    manual_preds.append(round(y_hat[0][0], 4))

# Keras predictions and loss
keras_preds = model.predict(X, verbose=0).flatten()
keras_loss_per_epoch = history.history['loss']

# ── TABLE 1: Predictions ─────────────────────────────────────────
print("=" * 76)
print("              PREDICTION COMPARISON (after 5 epochs)")
print("=" * 76)
print(f"  {'Student':<10} {'CGPA':<6} {'Profile':<10} {'Actual':<10} {'My Model':<12} {'Keras':<12} {'My Err':<10} {'Keras Err'}")
print("-" * 76)
for i in range(len(y)):
    m_err = abs(y[i] - manual_preds[i])
    k_err = abs(y[i] - keras_preds[i])
    print(f"  {i+1:<10} {X[i,0]:<6} {X[i,1]:<10} {y[i]:<10} {manual_preds[i]:<12.4f} {keras_preds[i]:<12.4f} {m_err:<10.4f} {k_err:.4f}")
print("=" * 76)

# ── TABLE 2: Loss per epoch ──────────────────────────────────────
print()
print("=" * 44)
print("         LOSS PER EPOCH COMPARISON")
print("=" * 44)
print(f"  {'Epoch':<8} {'My Model':<18} {'Keras'}")
print("-" * 44)
for i in range(5):
    print(f"  {i+1:<8} {manual_loss_per_epoch[i]:<18.4f} {keras_loss_per_epoch[i]:.4f}")
print("=" * 44)

# ── Summary ──────────────────────────────────────────────────────
avg_manual = np.mean([abs(y[i] - manual_preds[i]) for i in range(len(y))])
avg_keras  = np.mean([abs(y[i] - keras_preds[i])  for i in range(len(y))])
print(f"\n  Avg error — My model : {avg_manual:.4f}")
print(f"  Avg error — Keras      : {avg_keras:.4f}")
```
